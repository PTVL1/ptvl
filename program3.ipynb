```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

custom_colormap = ListedColormap([
    "#EABA0D",
    "#88DCE6",
    "#3FA6C5",
    "#F4D35E",
    "#EE964B"
])

colormap=custom_colormap,

text = "na disciplina de multimedia vamos aprender algumas coisas interessantes."

myStopwords = set(["Na", "de", "vamos", "algumas", "coisas","."])

nuvem = WordCloud (
    width = 1024, #px
    height = 768,
    background_color = "#B5B5B5",
    colormap = custom_colormap,
    max_words = 8,
    stopwords = myStopwords
).generate(text)

plt.figure(figsize=(10, 5))
plt.imshow(nuvem, interpolation = "bilinear")
plt.axis("off")
plt.show() # mostra a figura

nuvem.to_file("cloudOfWords.png")
```

```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# --- Personalização ---
texto = """
Na disciplina de multimédia vamos aprender programação, design,
imagem, vídeo, som, criatividade e muitas coisas interessantes.
"""

nome_ficheiro = "nuvem_multimedia.png"
cor_fundo = "#1E1E2E"
titulo = "O que vamos aprender em Multimédia?"

cores = ListedColormap([
    "#FFD166",  # amarelo
    "#06D6A0",  # verde-água
    "#118AB2",  # azul
    "#EF476F",  # rosa
    "#F8F9FA"   # branco
])

# Adiciona palavras que não devem aparecer
stopwords_personalizadas = set(STOPWORDS)
stopwords_personalizadas.update([
    "na", "de", "vamos", "algumas", "coisas", "e", "em"
])

# --- Criação da nuvem ---
nuvem = WordCloud(
    width=1400,
    height=800,
    background_color=cor_fundo,
    colormap=cores,
    max_words=30,
    stopwords=stopwords_personalizadas,
    contour_width=2,
    contour_color="#FFFFFF",
    prefer_horizontal=0.8,
    random_state=42
).generate(texto)

# --- Mostrar e guardar ---
plt.figure(figsize=(14, 7))
plt.imshow(nuvem, interpolation="bilinear")
plt.title(titulo, fontsize=20, fontweight="bold", color="#333333", pad=18)
plt.axis("off")
plt.tight_layout()
plt.show()

nuvem.to_file(nome_ficheiro)
print(f"Nuvem guardada em: {nome_ficheiro}")
```
